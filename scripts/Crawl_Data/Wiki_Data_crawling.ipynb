```python
from llama_index.core.node_parser import SentenceSplitter, SemanticSplitterNodeParser
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

if torch.cuda.is_available():
    device_type = "cuda"
else:
    device_type = "cpu"

# 1. Khởi tạo Model Embedding (như cũ)
embed_model = HuggingFaceEmbedding(
    model_name="dangvantuan/vietnamese-embedding",
    device=device_type,
    max_length=256
)
base_splitter = SentenceSplitter(chunk_size=256, chunk_overlap=0)

# 3. Đưa base_splitter vào trong SemanticSplitter
splitter = SemanticSplitterNodeParser(
    buffer_size=1,
    breakpoint_percentile_threshold=95,
    embed_model=embed_model,
    # QUAN TRỌNG: Dùng hàm split của base_splitter để đảm bảo đầu vào an toàn
    sentence_splitter=base_splitter.split_text
)


try:
    nodes = splitter.get_nodes_from_documents(documents)
    print(f"Thành công! Số lượng nodes: {len(nodes)}")
    if len(nodes) > 0:
        print("Node đầu tiên:", nodes[1].text[:300])
except Exception as e:
    print(f"Lỗi: {e}")
```

```python
def semantic_splitting_v2(sentences, embed_model, threshold=0.6):
    from sklearn.metrics.pairwise import cosine_similarity

    chunks = [[sentences[0]]]
    sequential_embedding = []

    for i in range(1, len(sentences)):
        sentence_i = sentences[i]
        embedding_i = embed_model.encode(sentence_i)  # Giữ nguyên vector

        if len(sequential_embedding) == 0:
            sentence_before_i = sentences[i-1]
            embedding_before_i = embed_model.encode(sentence_before_i)
            sequential_embedding.append(embedding_before_i)

        sequential_embedding.append(embedding_i)

        # Tính cosine similarity giữa 2 vectors
        sim_score = cosine_similarity(
            [sequential_embedding[-2]],
            [sequential_embedding[-1]]
        )[0][0]

        if sim_score >= threshold:
            chunks[-1].append(sentence_i)  # Bỏ dấu ngoặc vuông
        else:
            chunks.append([sentence_i])

        del sequential_embedding[0]

    chunks = [' '.join(chunk) for chunk in chunks]
    return chunks
```

```python
#function splitting
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def semantic_splitting_v2(sentences, embed_model, threshold=0.6):
    # 1. Xử lý ngoại lệ đầu vào
    if not sentences:
        return []

    # 2. Encode toàn bộ sentences một lần (Batch processing) -> Tăng tốc độ
    # convert_to_numpy=True để đảm bảo tính toán được với numpy/sklearn
    embeddings = embed_model.encode(sentences, convert_to_tensor=False)

    # 3. Khởi tạo
    chunks = [[sentences[0]]] # Bắt đầu chunk đầu tiên với câu đầu

    # 4. Duyệt qua các câu tiếp theo
    for i in range(1, len(sentences)):
        current_sentence = sentences[i]
        current_embedding = embeddings[i].reshape(1, -1)
        prev_embedding = embeddings[i-1].reshape(1, -1)

        # 5. Tính Cosine Similarity
        # Kết quả trả về ma trận [[score]], lấy [0][0] để ra số thực
        sim_score = cosine_similarity(prev_embedding, current_embedding)[0][0]

        # 6. Logic gom nhóm
        if sim_score >= threshold:
            # Nếu giống nhau (score cao), gộp vào chunk hiện tại
            chunks[-1].append(current_sentence)
        else:
            # Nếu khác nhau (score thấp), tạo chunk mới
            chunks.append([current_sentence])

    # 7. Gộp list các câu thành đoạn văn
    final_chunks = [' '.join(chunk) for chunk in chunks]

    return final_chunks
```
